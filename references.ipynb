```python
from pybtex.database import parse_file
from pybtex.errors import set_strict_mode

def clean_bib_urls(input_file, output_file):
    # Parse the bib file
    set_strict_mode(False)
    bib_data = parse_file(input_file)

    # Iterate through entries
    # bib_data.entries is a dictionary: {key: EntryObject}
    for key in bib_data.entries:
        entry = bib_data.entries[key]
        fields = entry.fields
        
        # Check if both 'doi' and 'url' exist in fields
        if 'doi' in fields and 'url' in fields:
            # Remove the url field
            del fields['url']

    # Write the modified database back to a file
    bib_data.to_file(output_file, bib_format='bibtex')

if __name__ == "__main__":
    clean_bib_urls('references-4.bib', 'references-2.bib')
```

```python
import bibtexparser


import bibtexparser
from bibtexparser.bwriter import BibTexWriter
from bibtexparser.bibdatabase import BibDatabase

def clean_bib_urls(input_file, output_file):
    # Load the bibtex file
    with open(input_file, 'r', encoding='utf-8') as bibfile:
        bib_database = bibtexparser.load(bibfile)

    # Iterate through each entry
    for entry in bib_database.entries:
        # Check if both 'doi' and 'url' fields exist
        if 'doi' in entry and 'url' in entry:
            # Remove the url field
            del entry['url']

    # Write the modified database to a new file
    writer = BibTexWriter()
    with open(output_file, 'w', encoding='utf-8') as bibfile:
        bibfile.write(writer.write(bib_database))

if __name__ == "__main__":
    input_path = 'references-3.bib'
    output_path = 'references_cleaned.bib'
    clean_bib_urls(input_path, output_path)
```
