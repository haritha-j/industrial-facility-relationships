```python
def model_inference(
    model, loader, device, calculate_score=False, return_coefficients=False
):
    predictor = model.eval()
    predictions_list, pcd_list, id_list, mean_list, norm_factor_list = (
        [],
        [],
        [],
        [],
        [],
    )
    with torch.no_grad():
        for j, data in tqdm(enumerate(loader), total=len(loader)):
            if return_coefficients:
                points, ids, mean, norm_factor = (
                    data["pointcloud"].to(device).float(),
                    data["id"].to(device),
                    data["mean"].to(device),
                    data["norm_factor"].to(device),
                )
            else:
                points, ids = data["pointcloud"].to(device).float(), data["id"].to(
                    device
                )
            points = points.transpose(2, 1)
            preds, _ = predictor(points)
            preds, ids, points = (
                preds.to(torch.device("cpu")),
                ids.to(torch.device("cpu")),
                points.to(torch.device("cpu")),
            )
            if return_coefficients:
                mean = mean.to(torch.device("cpu"))
                norm_factor = norm_factor.to(torch.device("cpu"))

            for i, pr in enumerate(preds):
                predictions_list.append(pr.numpy())
                pcd_list.append(points[i].numpy())
                id_list.append(ids[i].numpy())

                if return_coefficients:
                    mean_list.append(mean[i].numpy())
                    norm_factor_list.append(norm_factor[i].numpy())

        if return_coefficients:
            return (predictions_list, pcd_list, id_list, mean_list, norm_factor_list)
        return (predictions_list, pcd_list, id_list)
```

```python
def test(model, loader, device, criterion):
    losses = []
    predictor = model.eval()
    cloud_list = []
    label_list = []
    output_list = []
    predictions_list = []
    inputs_list = []
    id_list = []
    parameter_id = 0
    tot = 0
    count = 0

    for j, data in tqdm(enumerate(loader), total=len(loader)):
        inputs, labels, ids = (
            data["pointcloud"].to(device).float(),
            data["properties"].to(device),
            data["id"].to(device),
        )
        points, target, ids = (
            data["pointcloud"].to(device).float(),
            data["properties"].to(device),
            data["id"].to(device),
        )
        points = points.transpose(2, 1)
        outputs, _ = predictor(points)
        outputs = outputs.to(torch.device("cpu"))
        inputs = points.to(torch.device("cpu"))
        labels = target.to(torch.device("cpu"))
        ids = ids.to(torch.device("cpu"))
        # print(data['pointcloud'].size(), labels.size(), outputs.size())

        for i in range(outputs.size(0)):
            label_list.append(labels[i].numpy())
            id_list.append(ids[i].item())
            output_list.append(outputs[i][parameter_id].item())
            predictions_list.append(outputs[i].numpy())
            inputs_list.append(labels[i].numpy())
            cloud_list.append(inputs[i].numpy())
            ratio = (
                (labels[i][parameter_id] - outputs[i][parameter_id])
                / labels[i][parameter_id]
            ).item()
            # print('r', i+count, ids[i].item(), labels[i][parameter_id].item(), outputs[i][parameter_id].item(), ratio)
            tot += np.absolute(ratio)
            # print('l', labels[i][1].item(), outputs[i][1].item(), ((labels[i][1]-outputs[i][1])/labels[i][1]).item())

        count += outputs.size(0)
    print(tot / count)

    return predictions_list, inputs_list, label_list, output_list, id_list, cloud_list
```

```python
def save_debug_clouds(predictions_list, cloud_list, cat, index=0, output_prefix="debug"):
    """
    Generates and saves the predicted point cloud and the input point cloud for a specific example.
    
    Args:
        predictions_list: List of prediction arrays.
        cloud_list: List of input point cloud arrays (shape 3xN).
        cat: Category string (e.g., 'ibeam', 'cbeam').
        index: Index of the example to save.
        output_prefix: Prefix for the output filenames.
    """
    cuda = torch.device("cuda")
    
    # Select single example and prepare batch tensors
    pred = predictions_list[index]
    cloud = cloud_list[index] # Shape (3, N)
    
    # preds_t expects a batch
    preds_t = torch.tensor([pred], device=cuda, dtype=torch.float)
    
    # Generate point cloud from prediction
    # get_shape_cloud_tensor returns (Batch, Points, 3)
    generated_cloud_t = get_shape_cloud_tensor(preds_t, cat)
    generated_cloud_np = generated_cloud_t[0].detach().cpu().numpy()
    
    # Prepare input cloud for saving
    # cloud_list has (3, N), we need (N, 3) for PLY
    input_cloud_np = cloud.transpose()
    
    # Save Generated Cloud
    pcd_gen = o3d.geometry.PointCloud()
    pcd_gen.points = o3d.utility.Vector3dVector(generated_cloud_np)
    gen_filename = f"{output_prefix}_generated_{index}.ply"
    o3d.io.write_point_cloud(gen_filename, pcd_gen)
    print(f"Saved generated cloud to {gen_filename}")
    
    # Save Input Cloud
    pcd_in = o3d.geometry.PointCloud()
    pcd_in.points = o3d.utility.Vector3dVector(input_cloud_np)
    in_filename = f"{output_prefix}_input_{index}.ply"
    o3d.io.write_point_cloud(in_filename, pcd_in)
    print(f"Saved input cloud to {in_filename}")


save_debug_clouds(predictions_list, cloud_list, cat, index=4)
print((id_list[0]))
```

```python
#temp edit graph function

def plot_error_graph(
    data,
    label,
    x="Average bi-directional chamfer loss (mm)",
    y="No. of elements",
    max_val=None,
    negative=False,
):
    # filter top 99% to remove outliers
    sorted_data = np.sort(data)[::-1]*1000
    cap = int(len(data) / 100)
    filtered_data = sorted_data[cap:]
    norm = np.linalg.norm(filtered_data, ord=1)
    norm_data = filtered_data / norm
    print("avg", np.average(filtered_data), np.average(sorted_data), cap, filtered_data[0], filtered_data[-1])
    colour = "#ff7070"
    # draw graph
    fig = plt.figure(figsize=(12, 4))
    steps = 250
    if max_val is None:
        if negative:
            n, bins, _ = plt.hist(
                filtered_data,
                bins=np.arange(
                    filtered_data[-1],
                    filtered_data[0],
                    (filtered_data[0] - filtered_data[-1]) / steps,
                ),
                color=colour,
            )
        else:
            n, bins, _ = plt.hist(
                filtered_data,
                bins=np.arange(0, filtered_data[0], (filtered_data[0] - 0) / steps),
                color=colour,
            )
    else:
        n, bins, _ = plt.hist(
            filtered_data,
            bins=np.arange(0, max_val, (max_val - 0) / steps),
            color=colour,
        )
    mid = 0.5 * (bins[1:] + bins[:-1])
    plt.rcParams.update({"font.size": 20})
    
    ax = plt.gca()
    #ax.set_xlim([xmin, xmax])
    ax.set_ylim([0, 700])

    plt.xlabel(x, fontsize=20)
    plt.ylabel(y, fontsize=20)
    plt.title(label, fontsize=20)
    plt.errorbar(mid, n, yerr=0.01, fmt="none")
```

```python

# Parameter indices to penalise (prevent blowup during optimisation)
PARAM_PENALTY_INDICES = {
    "ibeam": [2, 3],   # web_thickness, flange_thickness
    "cbeam": [2, 3],
    "lbeam": [2, 3],
}

def chamfer_fine_tune(
    n_iter,
    step_size,
    preds,
    cloud,
    cat,
    blueprint,
    alpha=1.0,
    visualise=True,
    elbow_fix=True,
    robust=None,
    delta=0.1,
    bidirectional_robust=True,
    return_intermediate=False,
    k=1,
    direction_weight=None,
    loss_func= "chamfer",
    return_ifc=False,
    param_penalty_weight=1000.0,
):
    # prepare data on gpu and setup optimiser
    cuda = torch.device("cuda")
    preds_copy = copy.deepcopy(preds)
    scaled_original_preds = [scale_preds(pc.tolist(), cat) for pc in preds_copy]


    preds_t = torch.tensor(preds, requires_grad=True, device=cuda)
    cloud_t = torch.tensor(cloud, device=cuda)
    optimiser = torch.optim.Adam([preds_t], lr=step_size)

    # snapshot initial predictions for penalty reference
    initial_preds = preds_t.clone().detach()

    # check initial loss
    chamfer_loss_init, gen_cloud = get_chamfer_loss_tensor(
        preds_t, cloud_t, cat, reduce=False, return_cloud=True
    )
    gen_cloud = gen_cloud.detach().cpu().numpy()
    print("intial loss", torch.mean(chamfer_loss_init)*1000/2048, chamfer_loss_init)

    intermediate_results = []

    # iterative refinement with adam
    for i in tqdm(range(n_iter)):
        optimiser.zero_grad()
        if return_intermediate:
            intermediate_results.append(preds_t.clone().detach().cpu().numpy())

        if direction_weight is None:
            if loss_func == "chamfer":
                chamfer_loss = get_chamfer_loss_tensor(
                    preds_t,
                    cloud_t,
                    cat,
                    alpha=alpha,
                    robust=robust,
                    delta=delta,
                    bidirectional_robust=bidirectional_robust,
                )
            elif loss_func == "pair":
                chamfer_loss = get_pair_loss_tensor(preds_t, cloud_t, cat)
            elif loss_func == "emd":
                chamfer_loss = get_emd_loss_tensor(preds_t, cloud_t, cat)
            elif loss_func == "reverse":
                chamfer_loss = get_reverse_weighted_cd_tensor(preds_t, cloud_t, cat)
            elif loss_func == "balanced":
                chamfer_loss = get_balanced_chamfer_loss_tensor(preds_t, cloud_t, cat, k=k)
            elif loss_func == "symmetric":
                chamfer_loss = get_symmetric_chamfer_loss_tensor(preds_t, cloud_t, cat, k=k)
            elif loss_func == "infocd":
                chamfer_loss = get_infocd_loss_tensor(preds_t, cloud_t, cat)
        else:
            chamfer_loss = get_chamfer_loss_directional_tensor(
                preds_t,
                cloud_t,
                cat,
                alpha=alpha,
                robust=robust,
                delta=delta,
                bidirectional_robust=bidirectional_robust,
                k=k,
                direction_weight=direction_weight,
            )

        # parameter growth penalty
        if param_penalty_weight > 0.0 and cat in PARAM_PENALTY_INDICES:
            penalty_idx = PARAM_PENALTY_INDICES[cat]
            penalty = torch.mean(
                (preds_t[:, penalty_idx]- 0.005) ** 2
            )
            #print("cd", chamfer_loss, "param loss", param_penalty_weight * penalty)
            chamfer_loss = chamfer_loss + param_penalty_weight * penalty

        chamfer_loss.backward()
        optimiser.step()
        #print(i, "loss", chamfer_loss.detach().cpu().numpy())  # , "preds", preds_t)
        
        # measure CD at each itertation
        chamfer_loss_inter, gen_cloud = get_chamfer_loss_tensor(
            preds_t, cloud_t, cat, reduce=False, return_cloud=True
        )
        gen_cloud = gen_cloud.detach().cpu().numpy()
        print(torch.mean(chamfer_loss_inter).detach().cpu().numpy()*1000/2048)
        #print(chamfer_loss.detach().cpu().numpy()*1000/2048)  # , "preds", preds_t)

    # check final loss
    chamfer_loss, gen_cloud_mod = get_chamfer_loss_tensor(
        preds_t, cloud_t, cat, reduce=False, return_cloud=True
    )
    gen_cloud_mod = gen_cloud_mod.detach().cpu().numpy()
    #emd_loss = get_emd_loss_tensor(preds_t, cloud_t, cat)

    #print("final loss Chamfer", torch.mean(chamfer_loss).item(), "EMD", emd_loss.item())
    print("final losses", chamfer_loss)
    print("final loss Chamfer", torch.mean(chamfer_loss).item()*1000/2048)
    modified_preds = preds_t.detach().cpu().numpy()
    correct_id_list, incorrect_id_list = [], []

    # visualise
    if visualise:
        error_count = 0
        scaled_preds = [scale_preds(p.tolist(), cat) for p in modified_preds]
        visualisers = []
        cloud_visualisers = []
        vis_dict = {}

        for i, p in enumerate(scaled_preds):
            try:
                if not return_ifc:
                    v_orignal, _ = visualize_predictions(
                        [cloud[i].transpose(1, 0).tolist()],
                        cat,
                        [scaled_original_preds[i]],
                        blueprint,
                        visualize=(not return_ifc),
                    )
                v_modified, _ = visualize_predictions(
                    [None, None, cloud[i].transpose(1, 0).tolist()],
                    cat,
                    [scaled_preds[i]],
                    blueprint,
                    visualize=(not return_ifc),
                )
                visualisers.append(v_orignal)
                visualisers.append(v_modified)
                correct_id_list.append(id_list[i])
                #print("vis", i, id_list[i])
                vis_dict[str(id_list[i])] = (v_orignal, v_modified)
            except:
                print("error", i)
                error_count += 1
                v_orignal, _ = visualize_predictions(
                    [cloud[i].transpose(1, 0).tolist(), gen_cloud[i].tolist()],
                    cat,
                    [],
                    blueprint,
                    visualize=(not return_ifc),
                )
                v_modified, _ = visualize_predictions(
                    [
                        None,
                        gen_cloud_mod[i].tolist(),
                        cloud[i].transpose(1, 0).tolist(),
                    ],
                    cat,
                    [],
                    blueprint,
                    visualize=(not return_ifc),
                )
                incorrect_id_list.append(id_list[i])
                #print("vis", i, id_list[i])

                vis_dict[str(id_list[i])] = (v_orignal, v_modified)
                cloud_visualisers.append(v_orignal)
                cloud_visualisers.append(v_modified)

        #         return v_orignal,v_modified, modified_preds
        print("errors ", error_count)
        print("correct_id_list", correct_id_list)
        print("incorrect_id_list", incorrect_id_list)
        if return_ifc:
            return visualisers
        
        return cloud_visualisers, visualisers, modified_preds, vis_dict, correct_id_list
    else:
        if return_intermediate:
            return modified_preds, chamfer_loss, gen_cloud_mod, intermediate_results
        else:
            return modified_preds, chamfer_loss, gen_cloud_mod, chamfer_loss_init
```

### CLOI data extraction

```python
import json, os, math
import numpy as np


# ── helpers ──────────────────────────────────────────────────────────────────

def _decode_direction(preds, k):
    """Recover a normalised 3-D direction from 6 sin/cos trig values at index k.
    Matches geometry.get_direction_from_trig exactly."""
    d = [
        math.atan2(preds[k],     preds[k + 1]),
        math.atan2(preds[k + 2], preds[k + 3]),
        math.atan2(preds[k + 4], preds[k + 5]),
    ]
    n = math.sqrt(sum(v * v for v in d))
    return [v / n for v in d] if n > 1e-8 else d


def _ifc_local_frame(d, Z=(0., 0., 1.)):
    """Build the local coordinate frame used by CreateElbow / CreateBeam.
    IFC Axis2Placement3D:
        Z_local = direction   (Axis)
        X_local = cross(direction, Z_world)  (RefDirection)
        Y_local = Z_local × X_local
    Returns (X_local, Y_local) as unit numpy arrays.
    """
    d_np = np.array(d, dtype=float)
    Z_np = np.array(Z, dtype=float)
    X = np.cross(d_np, Z_np)
    if np.linalg.norm(X) < 1e-8:
        X = np.cross(d_np, np.array([0., 1., 0.]))
    X = X / np.linalg.norm(X)
    Y = np.cross(d_np, X)
    Y = Y / np.linalg.norm(Y)
    return X, Y


def _rodrigues(v, k, theta):
    """Rotate vector *v* around unit axis *k* by angle *theta* (rad)."""
    c, s = math.cos(theta), math.sin(theta)
    return v * c + np.cross(k, v) * s + k * float(np.dot(k, v)) * (1 - c)


def _pyvista_ref_dir(normal):
    """Image of X=(1,0,0) under the shortest-arc rotation from Z→*normal*.
    This is the reference direction (u = 0) that PyVista's ElbowModel uses
    when building the torus ring."""
    normal = np.asarray(normal, dtype=float)
    z = np.array([0., 0., 1.])
    x = np.array([1., 0., 0.])
    if np.allclose(normal, z, atol=1e-6):
        return x.copy()
    if np.allclose(normal, -z, atol=1e-6):
        return x.copy()                        # rotate_x(180) leaves X unchanged
    rot_axis = np.cross(z, normal)
    rot_axis = rot_axis / np.linalg.norm(rot_axis)
    rot_angle = math.acos(float(np.clip(np.dot(z, normal), -1, 1)))
    return _rodrigues(x, rot_axis, rot_angle)


# ── main extraction ──────────────────────────────────────────────────────────

def extract_params(preds, cat, norm_factor, mean):
    """Convert a single normalised prediction vector to a physical-units
    parameter dict compatible with the PyVista renderers."""
    preds = [float(v) for v in preds]
    nf = float(norm_factor)
    mx, my, mz = float(mean[0]), float(mean[1]), float(mean[2])

    def S(i):    return preds[i] * nf               # undo scale  → metres
    def P(i, m): return preds[i] * nf + m           # undo scale + translate → metres

    # ── ELBOW ────────────────────────────────────────────────────────────
    if cat in ("elbow", "bend"):
        # 1) Recover the raw IFC-level parameters
        r = S(0)                                        # minor radius  (m)
        x = S(1)                                        # axis_x in IFC local frame
        y = S(2)                                        # axis_y in IFC local frame
        p = np.array([P(3, mx), P(4, my), P(5, mz)])   # IFC placement origin (world)
        d = _decode_direction(preds, 8)                 # IFC Axis = local Z direction
        sweep_deg = math.degrees(math.atan2(preds[6], preds[7]))

        # 2) Build the IFC local coordinate frame
        X_loc, Y_loc = _ifc_local_frame(d)

        # 3) Torus centre = IFC origin + (x, y, 0) expressed in the local frame
        center = p + x * X_loc + y * Y_loc

        # 4) Torus plane normal = IFC revolution axis direction in world
        #    axis_dir in local 2D = (cos(theta), -sin(theta)) where theta = atan2(x,y)
        R = math.sqrt(x * x + y * y)
        if R > 1e-10:
            theta = math.atan2(x, y)
            ax, ay = math.cos(theta), -math.sin(theta)
            arc_normal = ax * X_loc + ay * Y_loc
            arc_normal = arc_normal / np.linalg.norm(arc_normal)
        else:
            arc_normal = np.array(d, dtype=float)
            R = max(R, 1e-6)

        # 5) Start angle – must match the PyVista ElbowModel convention:
        #    torus built in XY with Z-normal, then shortest-arc rotated Z→normal,
        #    then translated.  u = 0 reference = image of X=(1,0,0) under that rotation.
        #
        #    IFC sweep starts at the profile position (= IFC origin = p),
        #    which is at direction (p - center) from the torus centre.
        start_dir = p - center
        sn = np.linalg.norm(start_dir)
        if sn > 1e-10:
            start_dir = start_dir / sn

        e1 = _pyvista_ref_dir(arc_normal)           # u = 0° direction
        e2 = np.cross(arc_normal, e1)                # u = 90° direction
        start_angle = math.degrees(math.atan2(
            float(np.dot(start_dir, e2)),
            float(np.dot(start_dir, e1)),
        ))

        # Ensure positive sweep (PyVista requires sweep_angle > 0)
        if sweep_deg < 0:
            start_angle += sweep_deg
            sweep_deg = -sweep_deg
        start_angle %= 360.0

        return {
            "class": "elbow",
            "parameters": {
                "center_x":     float(center[0]),
                "center_y":     float(center[1]),
                "center_z":     float(center[2]),
                "normal_x":     float(arc_normal[0]),
                "normal_y":     float(arc_normal[1]),
                "normal_z":     float(arc_normal[2]),
                "major_radius": R,
                "minor_radius": abs(r),
                "start_angle":  start_angle,
                "sweep_angle":  abs(sweep_deg),
            }
        }

    # ── PIPE ─────────────────────────────────────────────────────────────
    elif cat == "pipe":
        d = _decode_direction(preds, 5)
        return {
            "class": "pipe",
            "parameters": {
                "pos_x":   P(2, mx),
                "pos_y":   P(3, my),
                "pos_z":   P(4, mz),
                "dir_x":   d[0],
                "dir_y":   d[1],
                "dir_z":   d[2],
                "length":  abs(S(1)),
                "radius":  abs(S(0)),
            }
        }

    # ── I-BEAM ───────────────────────────────────────────────────────────
    elif cat == "ibeam":
        d = _decode_direction(preds, 9)
        d_np = np.array(d, dtype=float)
        length = abs(S(5))

        # preds[6:9] = IFC placement origin = start (one end) of the beam extrusion.
        # IBeamModel.pos_* is the GEOMETRIC CENTRE, so shift by half length along axis.
        start = np.array([P(6, mx), P(7, my), P(8, mz)])
        center = start + d_np * (length / 2)

        # Use world-Z as the up direction. IBeamModel.generate_mesh applies
        # Gram-Schmidt to orthogonalise it w.r.t. the beam axis, so this is safe
        # for all non-vertical beams. Using cross(d, Z) (the old code) gave
        # the *side* direction, rotating the cross-section 90 degrees.
        return {
            "class": "ibeam",
            "parameters": {
                "pos_x":            float(center[0]),
                "pos_y":            float(center[1]),
                "pos_z":            float(center[2]),
                "dir_x":            d[0],
                "dir_y":            d[1],
                "dir_z":            d[2],
                "up_x":             0.0,
                "up_y":             0.0,
                "up_z":             1.0,
                "height":           abs(S(1)),   # preds[1] = depth  → IBeamModel height (H)
                "width":            abs(S(0)),   # preds[0] = width  → IBeamModel width  (W)
                "flange_thickness": abs(S(3)),
                "web_thickness":    abs(S(2)),
                "length":           length,
            }
        }

    else:
        raise ValueError(f"Unsupported category: {cat}")

# ── run extraction ────────────────────────────────────────────────────────────
output_base = "params"

for idx, (preds_row, elem_id) in enumerate(zip(modified_preds, id_list)):
    if elem_id not in correct_id_list:
        print("skipping")
        continue
    else:
        print(elem_id)

    nf   = norm_factor_list[idx] if "norm_factor_list" in dir() else 1.0
    mean = mean_list[idx]         if "mean_list"        in dir() else np.zeros(3)

    try:
        params = extract_params(preds_row, cat, nf, mean)
    except Exception as e:
        print(f"[{idx}] id={elem_id}  ERROR: {e}")
        continue

    # resolve element id (may be int array or plain int/str)
    if hasattr(elem_id, '__iter__'):
        eid = int(elem_id.flat[0])
    else:
        eid = int(elem_id)

    out_dir  = os.path.join(output_base, cat)
    os.makedirs(out_dir, exist_ok=True)
    out_path = os.path.join(out_dir, f"{eid}.json")
    with open(out_path, "w") as f:
        json.dump(params, f, indent=2)

print(f"Saved {len(correct_id_list)} '{cat}' parameter files to '{output_base}/{cat}/'")
```

```python
%autoreload 2
# morph a sphere into the shape of an input point cloud
# by optimising chamfer loss iteratively
# total points = num_points**2
def optimise_shape(src_pcd_tensor, tgt_pcd_tensor, iterations=5, learning_rate=0.01, loss_func= "chamfer"):

    cuda = torch.device("cuda")

    # optimise
    optimizer = torch.optim.Adam([tgt_pcd_tensor], lr=learning_rate)
    intermediate, losses, assingments = [], [], []
    chamferDist = ChamferDistance()
    assignments = []

    for i in tqdm(range(iterations)):
        optimizer.zero_grad()

        if loss_func == "chamfer":
            nn = chamferDist(
                src_pcd_tensor, tgt_pcd_tensor, bidirectional=True, return_nn=True)
            loss = torch.sum(nn[1].dists) + torch.sum(nn[0].dists)
            assignment = [nn[0].idx[:,:,0].detach().cpu().numpy(), nn[1].idx[:,:,0].detach().cpu().numpy()]
        elif loss_func == "emd":
            loss, assignment = calc_emd(tgt_pcd_tensor, src_pcd_tensor, 0.05, 50)
            assignment = assignment.detach().cpu().numpy()
        elif loss_func == "balanced":
            loss, assignment = calc_balanced_chamfer_loss_tensor(src_pcd_tensor, tgt_pcd_tensor, return_assignment=True, k=4)
        elif loss_func == "infocd":
            loss, assignment = calc_cd_like_InfoV2(src_pcd_tensor, tgt_pcd_tensor, return_assignment=True)
        elif loss_func == "direct":
            loss = calc_direct(src_pcd_tensor, tgt_pcd_tensor)
        else:
            print("unspecified loss")

        #print("a", assignment[0].shape)
        loss.backward()
        optimizer.step()
        print("iteration", i, "loss", loss.item())

        intermediate.append(tgt_pcd_tensor.clone())

    # calculate final chamfer loss
    dist = chamferDist(src_pcd_tensor, tgt_pcd_tensor, bidirectional=True)
    emd_loss, _ = calc_emd(tgt_pcd_tensor, src_pcd_tensor, 0.05, 50)
    print("final chamfer dist", dist.item(), "emd", emd_loss.item())

    # save assignments for analysis
#     if measure_consistency:
#         with open("sphere/assignments_" + loss_func + ".pkl", "wb") as f:
#             pickle.dump(assingments, f)

    intermediate = torch.stack(intermediate)
    return intermediate
```

```python
# compare losses
def evaluate_autoencoder_results(inp_list, rec_list, loss_funcs, batch_size=None):
    cuda = torch.device("cuda")
    if batch_size == None:
        batch_size = len(inp_list)

    # create empty dict
    all_losses = {func: [] for func in loss_funcs}

    # split into batches
    for i in tqdm(range(math.ceil(len(inp_list) / batch_size))):
        a = i * batch_size
        b = min(a + batch_size, len(inp_list))
        # print(a,b)

        # compute losses
        inp_tensor = torch.tensor(inp_list[a:b], device=cuda)
        rec_tensor = torch.tensor(rec_list[a:b], device=cuda)
        losses = calculate_3d_loss(inp_tensor, rec_tensor, loss_funcs)

        # sum losses
        for k, v in losses.items():
            all_losses[k].append(v)

    # average losses
    for k in all_losses:
        all_losses[k] = np.average(np.array(all_losses[k]))
    return all_losses
```

```python
# produce density colourmaps that are normalised with their pairs


# get densities
def get_density(clouds):
    # compute nearest neighbours to calculated density
    clouds = torch.tensor(clouds, device="cuda")
    chamferDist = ChamferDistance()
    nn = chamferDist(clouds, clouds, bidirectional=False, return_nn=True, k=32)

    density = torch.mean(nn[0].dists[:, :, 1:], dim=2)
    eps = 0.00001
    density = 1 / (density + eps)
    return density


# normalise for each example across prediction sets
def normalise_densities(density_sets):
    densities = torch.stack(density_sets)
    highs, lows = torch.max(densities, 2).values, torch.min(densities, 2).values
    highs, lows = torch.max(highs, 0).values, torch.min(lows, 0).values
    # print(densities.shape, highs.shape)

    # highs = torch.reshape(highs, densities.shape)
    highs = highs.unsqueeze(0).unsqueeze(-1)
    highs = highs.expand(densities.shape[0], densities.shape[1], densities.shape[2])
    lows = lows.unsqueeze(0).unsqueeze(-1)
    lows = lows.expand(densities.shape[0], densities.shape[1], densities.shape[2])
    diff = highs - lows
    densities = (densities - lows) / diff

    return densities[0], densities[1], densities[2], densities[3]


# represent density with colour and combine with point cloud
def get_coloured_clouds(clouds, density, colormap_name="plasma_r"):
    density = density.detach().cpu().numpy()
    colours = np.zeros((density.shape[0], density.shape[1], 4))
    colormap = plt.get_cmap(colormap_name)

    for i, cloud in enumerate(density):
        for j, pt in enumerate(cloud):
            colours[i, j] = colormap(pt)

    #     clouds = clouds.detach().cpu().numpy()
    colours = colours[:, :, :3]
    pcds = []

    for i, cl in enumerate(clouds):
        pcd = o3d.geometry.PointCloud()
        pcd.points = o3d.utility.Vector3dVector(cl)
        pcd.colors = o3d.utility.Vector3dVector(colours[i])
        pcds.append(pcd)

    return pcds, colours
```

```python
# rescale and move obj file

def scale_obj(input_file, output_file, scale):
    with open(input_file, 'r') as f:
        lines = f.readlines()

    scaled_vertices = []

    for line in lines:
        if line.startswith('v '):
            parts = line.split()
            x = float(parts[1]) * scale
            y = float(parts[2]) * scale - 0.5
            z = float(parts[3]) * scale
            scaled_vertices.append((x, y, z))
        else:
            scaled_vertices.append(line)

    with open(output_file, 'w') as f:
        for item in scaled_vertices:
            if isinstance(item, tuple):
                f.write("v {} {} {}\n".format(item[0], item[1], item[2]))
            else:
                f.write(item)


# Example usage:
input_file = "scaled_lamp3.obj"
output_file = "scaled_lamp4.obj"
scale = 3


scale_obj(input_file, output_file, scale)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from mpl_toolkits.mplot3d.art3d import Poly3DCollection

def create_swept_volume_diagram():
    fig = plt.figure(figsize=(10, 8))
    ax = fig.add_subplot(111, projection='3d')

    # 1. Define the 2D shape (Circle)
    theta = np.linspace(0, 2 * np.pi, 50)
    radius = 0.5
    circle_x = radius * np.cos(theta)
    circle_y = radius * np.sin(theta)

    # 2. Define the path (a gentle curve)
    z_path = np.linspace(0, 5, 100)
    # Simple curve: a slight spiral/arc
    path_x = 0.5 * np.sin(z_path * 0.5)
    path_y = 0.5 * (1 - np.cos(z_path * 0.5))

    # Plot the path
    ax.plot(path_x, path_y, z_path, 'r--', lw=2, label='Path (Trajectory)')

    # 3. Create the swept surface (Tube)
    # Generate the vertices for the surface
    X = []
    Y = []
    Z = []
    for i in range(len(z_path)):
        X.append(circle_x + path_x[i])
        Y.append(circle_y + path_y[i])
        Z.append(np.full_like(circle_x, z_path[i]))

    X = np.array(X)
    Y = np.array(Y)
    Z = np.array(Z)

    # Plot the swept volume surface
    ax.plot_surface(X, Y, Z, color='skyblue', alpha=0.4, edgecolor='none')
    
    # 4. Show the cross-section "sweeping"
    # Plot ribs at specific intervals
    indices = [0, 25, 50, 75, 99]
    colors = ['green', 'blue', 'blue', 'blue', 'navy']
    labels = ['Original 2D Area', None, None, None, 'Final Swept Position']
    
    for idx, i in enumerate(indices):
        ax.plot(X[i], Y[i], Z[i], color=colors[idx], lw=2, label=labels[idx] )

    # To create a solid "cap", we can plot a surface for the disk
    r_grid = np.linspace(0, radius, 10)
    theta_grid = np.linspace(0, 2*np.pi, 30)
    R, T = np.meshgrid(r_grid, theta_grid)
    
    # Bottom cap
    cap_x0 = R * np.cos(T) + path_x[0]
    cap_y0 = R * np.sin(T) + path_y[0]
    cap_z0 = np.full_like(cap_x0, z_path[0])
    ax.plot_surface(cap_x0, cap_y0, cap_z0, color='navy', alpha=0.6)

    # Top cap
    cap_x1 = R * np.cos(T) + path_x[-1]
    cap_y1 = R * np.sin(T) + path_y[-1]
    cap_z1 = np.full_like(cap_x1, z_path[-1])
    ax.plot_surface(cap_x1, cap_y1, cap_z1, color='navy', alpha=0.6)

    # Labels and formatting
    ax.set_xlabel('X', fontsize=14)
    ax.set_ylabel('Y', fontsize=14)
    ax.set_zlabel('Z (Extrusion)', fontsize=14)
    ax.set_title('Swept Area Solid Representation\n(2D Circle extruded along a 3D path)')
    ax.legend(fontsize=15)
    
    # Set axis limits for better viewing
    ax.set_xlim(-1, 2)
    ax.set_ylim(-1, 2)
    ax.set_zlim(0, 5)
    ax.tick_params(axis='both', which='major', labelsize=12)
    
    ax.view_init(elev=25, azim=30)

    plt.show()
    plt.close()

create_swept_volume_diagram()
```

```python
# CLOI evaluation against annotated parameters
import os
import json
import numpy as np

def calculate_pipe_errors(annotations_dir, predictions_dir, unit_is_meters=True):
    results = {}
    annotation_files = [f for f in os.listdir(annotations_dir) if f.endswith('.json')]
    
    for filename in annotation_files:
        pred_path = os.path.join(predictions_dir, filename)
        if not os.path.exists(pred_path):
            continue
            
        anno_path = os.path.join(annotations_dir, filename)
        with open(anno_path, 'r') as f:
            anno_data = json.load(f)
            
        if anno_data.get("skip", False):
            continue
            
        with open(pred_path, 'r') as f:
            pred_data = json.load(f)
            
        a_params = anno_data["parameters"]
        p_params = pred_data["parameters"]
        
        # 1. Radius percentage error
        a_rad = a_params["radius"]
        p_rad = p_params["radius"]
        rad_err_pct = (abs(p_rad - a_rad) / a_rad * 100.0) if a_rad != 0 else 0.0
            
        # 2. Length percentage error (Added)
        a_len = a_params["length"]
        p_len = p_params["length"]
        len_err_pct = (abs(p_len - a_len) / a_len * 100.0) if a_len != 0 else 0.0

        # 3. Position error in mm
        a_pos = np.array([a_params["pos_x"], a_params["pos_y"], a_params["pos_z"]])
        p_pos = np.array([p_params["pos_x"], p_params["pos_y"], p_params["pos_z"]])
        pos_err = np.linalg.norm(a_pos - p_pos)
        pos_err_mm = pos_err * 1000.0 if unit_is_meters else pos_err
            
        # 4. Axis error in degrees
        a_dir = np.array([a_params["dir_x"], a_params["dir_y"], a_params["dir_z"]])
        p_dir = np.array([p_params["dir_x"], p_params["dir_y"], p_params["dir_z"]])
        a_dir_norm = a_dir / (np.linalg.norm(a_dir) or 1e-8)
        p_dir_norm = p_dir / (np.linalg.norm(p_dir) or 1e-8)
        
        # Handle pipe symmetry (lowest angle between lines)
        dot_product = np.abs(np.clip(np.dot(a_dir_norm, p_dir_norm), -1.0, 1.0))
        angle_deg = np.degrees(np.arccos(dot_product))
        
        _id = filename.split('.')[0]
        results[_id] = {
            "radius_error_pct": rad_err_pct,
            "length_error_pct": len_err_pct,
            "position_error_mm": pos_err_mm,
            "axis_error_deg": angle_deg
        }
        
    return results

def calculate_average_pipe_errors(errors_dict):
    if not errors_dict:
        print("No valid pipe matches found.")
        return None
        
    rad_errs = [data["radius_error_pct"] for data in errors_dict.values()]
    len_errs = [data["length_error_pct"] for data in errors_dict.values()]
    pos_errs = [data["position_error_mm"] for data in errors_dict.values()]
    axis_errs = [data["axis_error_deg"] for data in errors_dict.values()]
    
    averages = {
        "mean_radius_error_pct": np.mean(rad_errs),
        "mean_length_error_pct": np.mean(len_errs),
        "mean_position_error_mm": np.mean(pos_errs),
        "mean_axis_error_deg": np.mean(axis_errs),
        "total_evaluated": len(errors_dict)
    }
    
    print(f"--- Average Errors across {averages['total_evaluated']} pipes ---")
    print(f"Mean Radius Error: {averages['mean_radius_error_pct']:.2f}%")
    print(f"Mean Length Error: {averages['mean_length_error_pct']:.2f}%")
    print(f"Mean Pos Error:    {averages['mean_position_error_mm']:.2f} mm")
    print(f"Mean Axis Error:   {averages['mean_axis_error_deg']:.2f} deg")
    
    return averages

# --- Usage Example ---
annotations_folder = "../experiments/param_annotate/data/pipe/params/"
predictions_folder =  "../experiments/param_annotate/src/output/pipe/"


errors = calculate_pipe_errors(annotations_folder, predictions_folder)
average_metrics = calculate_average_pipe_errors(errors)


# errors = calculate_pipe_errors(annotations_folder, predictions_folder)

# for pipe_id, errs in errors.items():
#     print(f"ID {pipe_id}:")
#     print(f"  Radius Error: {errs['radius_error_pct']:.2f}%")
#     print(f"  Pos Error:    {errs['position_error_mm']:.2f} mm")
#     print(f"  Axis Error:   {errs['axis_error_deg']:.2f} deg\n")
```

```python
import os
import json
import numpy as np

def calculate_ibeam_errors(annotations_dir, predictions_dir, unit_is_meters=True):
    results = {}
    
    # Get all json files in annotations directory
    annotation_files = [f for f in os.listdir(annotations_dir) if f.endswith('.json')]
    
    for filename in annotation_files:
        pred_path = os.path.join(predictions_dir, filename)
        
        # Check if the same id exists in predictions
        if not os.path.exists(pred_path):
            continue
            
        anno_path = os.path.join(annotations_dir, filename)
        
        with open(anno_path, 'r') as f:
            anno_data = json.load(f)
            
        # Check if we need to skip this annotation
        if anno_data.get("skip", False):
            continue
            
        with open(pred_path, 'r') as f:
            pred_data = json.load(f)
            
        a_params = anno_data["parameters"]
        p_params = pred_data["parameters"]
        
        # 1. Length percentage error
        a_length = a_params["length"]
        p_length = p_params["length"]
        if a_length == 0:
            len_err_pct = 0.0 # Prevent division by zero
        else:
            len_err_pct = (abs(p_length - a_length) / a_length) * 100.0
            
        # 2. Position error in mm
        a_pos = np.array([a_params["pos_x"], a_params["pos_y"], a_params["pos_z"]])
        p_pos = np.array([p_params["pos_x"], p_params["pos_y"], p_params["pos_z"]])
        
        # Euclidean distance
        pos_err = np.linalg.norm(a_pos - p_pos)
        pos_err_mm = pos_err * 1000.0 if unit_is_meters else pos_err
            
        # 3. Error between the main axes (dir) in degrees
        a_dir = np.array([a_params["dir_x"], a_params["dir_y"], a_params["dir_z"]])
        p_dir = np.array([p_params["dir_x"], p_params["dir_y"], p_params["dir_z"]])
        
        # Normalize the direction vectors
        a_dir_norm = a_dir / (np.linalg.norm(a_dir) or 1e-8)
        p_dir_norm = p_dir / (np.linalg.norm(p_dir) or 1e-8)
        
        # Dot product (absolute value due to I-beam symmetry along length)
        dot_product = np.abs(np.clip(np.dot(a_dir_norm, p_dir_norm), -1.0, 1.0))
        angle_rad = np.arccos(dot_product)
        angle_deg = np.degrees(angle_rad)
        
        _id = filename.split('.')[0]
        results[_id] = {
            "length_error_pct": len_err_pct,
            "position_error_mm": pos_err_mm,
            "axis_error_deg": angle_deg
        }
        
    return results

def calculate_average_ibeam_errors(errors_dict):
    if not errors_dict:
        print("No valid I-beam matches found.")
        return None
        
    len_errors = [data["length_error_pct"] for data in errors_dict.values()]
    pos_errors = [data["position_error_mm"] for data in errors_dict.values()]
    axis_errors = [data["axis_error_deg"] for data in errors_dict.values()]
    
    averages = {
        "mean_length_error_pct": np.mean(len_errors),
        "mean_position_error_mm": np.mean(pos_errors),
        "mean_axis_error_deg": np.mean(axis_errors),
        "total_evaluated": len(errors_dict)
    }
    
    print(f"--- Average Errors across {averages['total_evaluated']} I-beams ---")
    print(f"Mean Length Error: {averages['mean_length_error_pct']:.2f}%")
    print(f"Mean Pos Error:    {averages['mean_position_error_mm']:.2f} mm")
    print(f"Mean Axis Error:   {averages['mean_axis_error_deg']:.2f} deg")
    
    return averages


ibeam_annotations_folder = "../experiments/param_annotate/data/ibeam/params/"
ibeam_predictions_folder = "../experiments/param_annotate/src/output/ibeam/"

ibeam_errors = calculate_ibeam_errors(ibeam_annotations_folder, ibeam_predictions_folder)
ibeam_averages = calculate_average_ibeam_errors(ibeam_errors)
```
